```python
#lets carry out eda on an inbuilt dataset in seaborn library called 'titanic'
#importing the libraries
import seaborn as sns
```

```python
#lets load the dataset  
titanic = sns.load_dataset('titanic')
```

```python
#show the first 5 rows of the dataset
titanic.head()
```

```python
#create visualization to show the distribution of age in the dataset
```

```python
# plotting the distribution of age using seaborn's distplot
sns.distplot(titanic['age'].dropna(), kde=False, bins=30)
```

```python
#plot a boxplot to show the distribution ofage with respect to the class of the passengers
sns.boxplot(x='class', y='age', data=titanic)
```
